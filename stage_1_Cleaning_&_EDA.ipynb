# Quality Report 

```python
# function makes column wise report 

def column_report(df):
    report = pd.DataFrame({
        'Column': df.columns,
        'Data_Type': df.dtypes.astype(str).values,
        'Total_Entries': df.count().values,
        'Null_Count': df.isnull().sum().values,
        'Null_Percentage': (
            df.isnull().mean().values * 100
        ).round(2),
        'Unique_Values': df.nunique().values,
        'Sample_Values': [
            df[col].dropna().unique()[:5].tolist()
            for col in df.columns
        ]
    })

    return report


def generate_reports(df1, df2):
    report1 = column_report(df1)
    report2 = column_report(df2)

    return report1,report2
```

```python
def df_and_column_cleaning(df):
    df=df
    #clean column naming
    df.columns = [columns.title().strip().replace(" ","_") for columns in df.columns]
    # colun selection
    str_col=df.select_dtypes(include=str).columns
    numeric_col=df.select_dtypes(include=[int,float]).columns

    #for str column cleaning 
    for col in str_col:
        df[col]=df[col].str.strip().str.title()

    return df

    
```

```python
df_profile_clean=df_and_column_cleaning(df_profile)
df_meta_clean=df_and_column_cleaning(df_meta)
```

# Restructuring Data

```python


meta_year_cols = [
    col for col in df_meta_clean.columns
    if col.rsplit('_', 1)[-1].isdigit()
]

meta_year_cols
```

```python
#converting wide to long format
meta_long = df_meta_clean.melt(
    id_vars='Ee_Ref',
    value_vars=meta_year_cols,
    var_name='attribute_year',
    value_name='value'
)
meta_long
```

```python
#extracting year from attribute

meta_long['year'] = (
    meta_long['attribute_year']
    .str.extract(r'(\d{4})$')
)
meta_long
```

```python
#extracting attribute 

meta_long['attribute'] = (
    meta_long['attribute_year']
    .str.replace(r'_\d{4}$', '', regex=True)
)
meta_long.head(10)
```

```python
meta_long["year"]=meta_long["year"].astype(int)
```

```python
#converting into proper format


meta_final = meta_long.pivot(
    index=['Ee_Ref', 'year'],
    columns='attribute',
    values='value'
).reset_index()

meta_final
```

```python
#checking duplicate
meta_final.duplicated(
    subset=['Ee_Ref', 'year']
).sum()
```

```python
#checking avg and outliers 

numeric_col=df_profile_clean.select_dtypes(include=[int,float]).columns
numeric_col=numeric_col.drop("Submission_Year")

df_coulmns_report=[]
for col in numeric_col:
    detail={}
    #avg min and max detection
    avg=df_profile_clean[col].mean().round(2)
    minimum=df_profile_clean[col].min()
    maximum=df_profile_clean[col].max()
    detail["column_name"]=col
    detail["average"]=avg
    detail["minimum_value"]=minimum
    detail["maximum_value"]=maximum


    #outliers deetction

    q1=df_profile_clean[col].quantile(.25)
    q3=df_profile_clean[col].quantile(.75)

    IQR=q3-q1

    lower_bound=q1-1.5*IQR
    upper_bound=q3+1.5*IQR

    detail["Q1"]=q1
    detail["Q3"]=q3
    detail["IQR"]=IQR
    detail["lower_bound"]=lower_bound
    detail["upper_bound"]=upper_bound
    

    df_coulmns_report.append(detail)


columns_report_df=pd.DataFrame(df_coulmns_report)
columns_report_df.to_csv(Path("reports")/"ee_profile_avg_and_outliers_report.csv",index=False)
```

```python
#zero and non zero count for every demographic 
zero_summary = pd.DataFrame({
    'Zero_Count': (final_df[demographic_cols] == 0).sum(),
    'Non_Zero_Count': (final_df[demographic_cols] > 0).sum()
})

zero_summary['Zero_Percentage'] = (
    zero_summary['Zero_Count'] / len(final_df) * 100
)

zero_summary['Non_Zero_Percentage'] = (
    zero_summary['Non_Zero_Count'] / len(final_df) * 100
)

zero_summary=zero_summary.reset_index(names="Demographic")
zero_summary=zero_summary.melt(id_vars="Demographic",value_name="Count",var_name="Metric")
zero_summary
```

```python
#calculation yearly percentage share 
yearly_percentage = yearly_workforce.copy()

yearly_percentage['Total_Workforce'] = yearly_percentage[demographic_cols].sum(axis=1)

for col in demographic_cols:
    yearly_percentage[col] = (
        yearly_percentage[col] /
        yearly_percentage['Total_Workforce'] * 100
    ).round(2)

yearly_percentage
```

```python
occupation_workforce = final_df.groupby('Occupation_Level')[demographic_cols].sum()

occupation_workforce['Total_Workforce'] = (occupation_workforce[demographic_cols].sum(axis=1))

occupation_workforce = occupation_workforce.sort_values('Total_Workforce',ascending=False)

occupation_workforce
```

```python
occupation_chart = occupation_workforce.reset_index()

plt.figure(figsize=(54,20))

ax=sns.barplot(
    data=occupation_chart,
    x='Occupation_Level',
    y='Total_Workforce',
    hue='Occupation_Level',
    palette='Set2',
    legend=False
)
for cont in ax.containers:
    ax.bar_label(cont,padding=2,fmt="%d",fontsize=20)
plt.title('Total Workforce by Occupational Level', fontsize=32)
plt.xlabel('Occupational Level', fontsize=11)
plt.ylabel('Total Employee Count', fontsize=11)
plt.yticks(fontsize=20,color="g")

plt.xticks(rotation=45,ha="right", fontsize=20, color="red" )
plt.tight_layout()
plt.grid(axis='y', alpha=0.3)


plt.savefig(
    visual_dir/ 'Total Workforce by Occupational Level.png',
    dpi=300,
    bbox_inches='tight'
)




plt.show()
```

```python
occupation_percentage = (occupation_workforce[demographic_cols].div(
    occupation_workforce[demographic_cols].sum(axis=1),
    axis=0
) * 100).round(2)
occupation_percentage
```

```python
occupation_percentage_plot = occupation_percentage.reset_index()

occupation_percentage_plot = occupation_percentage_plot.melt(
    id_vars='Occupation_Level',
    var_name='Demographic',
    value_name='Percentage'
)

plt.figure(figsize=(54, 14))

ax=sns.barplot(
    data=occupation_percentage_plot,
    x='Occupation_Level',
    y='Percentage',
    hue='Demographic',
    palette='tab10'
)
for cont in ax.containers:
    ax.bar_label(cont,padding=2,fontsize=12,color="brown")

plt.title('Demographic Composition Across Occupational Levels', fontsize=16)
plt.xlabel('Occupational Level', fontsize=11)
plt.ylabel('Workforce Percentage (%)', fontsize=11)

plt.legend(title='Demographic', bbox_to_anchor=(1, 1), loc='upper left')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.grid(axis='y', alpha=0.3)


plt.savefig(
    visual_dir/ 'Demographic Composition Across Occupational Levels.png',
    dpi=300,
    bbox_inches='tight'
)


plt.show()
```
